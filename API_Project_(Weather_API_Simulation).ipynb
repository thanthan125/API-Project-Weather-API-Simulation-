```python
# ==================================
# WEATHER API PROJECT
# ==================================

!pip install requests pandas --quiet

import requests
import pandas as pd

# Using public open-meteo API (no key needed)
url = "https://api.open-meteo.com/v1/forecast?latitude=17.3850&longitude=78.4867&daily=temperature_2m_max&timezone=auto"

response = requests.get(url)
data = response.json()

dates = data["daily"]["time"]
temps = data["daily"]["temperature_2m_max"]

df = pd.DataFrame({
    "Date": dates,
    "Max_Temperature": temps
})

df.to_csv("weather_data.csv", index=False)

print("Weather data collected:")
df.head()
```
